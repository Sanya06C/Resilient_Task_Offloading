# Reproduce the audited paper tables
This notebook averages nested EH episodes within independent model seeds. It reads the corrected released records, regenerates tables and figures, and verifies evidence integrity. It does not retrain models or run expensive simulations.

```python
from pathlib import Path
import os, subprocess, sys, json
root = Path.cwd()
if root.name == "publication":
    root = root.parent
assert (root / "publication").is_dir(), "Run from the repository root or publication directory"
os.chdir(root)
for script in ["recompute_statistics.py", "build_tables.py", "make_standalone.py", "check_evidence.py"]:
    subprocess.run([sys.executable, str(root / "publication" / script)], check=True)
```

```python
results = json.loads((root / "publication/evidence/eh_seed_statistics.json").read_text())
for stress, metrics in results.items():
    print(stress, metrics["terminal_mean_final_soc"]["paired_difference"])
```
